```python
from transformers import pipeline
```

```python
!pip install arxiv
import arxiv
import pandas as pd
```

```python
# query to fetch ai-related papers
query = "AI or Artificial Intelligence OR Machine Learning"
client = arxiv.Client() # Initialize arxiv client
search = arxiv.Search(query=query, max_results=10, sort_by=arxiv.SortCriterion.SubmittedDate)

# fetch papers
papers = []
for result in client.results(search):
  papers.append({
      'published': result.published,
      'title': result.title,
      'abstract': result.summary, # Corrected: use result.summary instead of result.abstract
      'categories': result.categories
  })

# convert to dataframe (moved outside the loop)
df = pd.DataFrame(papers) # Corrected typo: DatFrame -> DataFrame
pd.set_option('display.max_colwidth', None) # Corrected option name: 'display.max_colwidths' -> 'display.max_colwidth'
df.head()
```

```python
abstract = df['abstract'][0]
summarizer = pipeline("text-generation", model="facebook/bart-large-cnn")
```

```python
summarization_result = summarizer(abstract, max_new_tokens=100)
print(summarization_result[0]['generated_text'])
```
