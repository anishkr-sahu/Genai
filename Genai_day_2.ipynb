```python
# Bag of words : it counts the frequency of occurrence of unique words in the document
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({"text": ["people watch dswithbappy",
                            "dswithbappy watch dswithbappy",
                            "people write comment",
                            "dswithbappy write comment"],
                   "output": [1,1,0,0]})
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
bow = cv.fit_transform(df["text"])
print(cv.vocabulary_)
```

```python
bow.toarray()
```

```python
# n- grams : take n words as a one word
cv = CountVectorizer(ngram_range=(2,2))
bow = cv.fit_transform(df["text"])
print(cv.vocabulary_)
```

```python
bow.toarray()
```

```python
# word2Vec
sentences = [
    ["nlp", "is", "Interesting"],
    ["machine", "learning","is","powerful"],
    ["deep","learning","is","advanced"],
    ["nlp","and","machine","learning"],
    ["i","love","studying","nlp"],
    ["students","prefer","machine","learning"],
    ["nlp","helps","in","text","processing"],
    ["learning","new","'topics","is","fun"],
    ["artificial","Intelligence","is","future"]
]
```

```python
!pip install gensim
```

```python
from gensim.models import Word2Vec
# define model & training
model = Word2Vec(
    sentences,
    vector_size= 50,
    window=3,
    min_count=1,
    workers=700
)
print("vector for nlp")
print(model.wv['nlp'])
print("\nwords similar to nlp")
print(model.wv.most_similar('nlp'))
```

```python
# vocabulary
print("Vocabulary: ")
print(model.wv.index_to_key)
```

```python
# sentence to vector
import numpy as np
def sent_vec(sentence):
  words = sentence.lower().split()
  vectors= []
  for word in words:
    if word in model.wv:
      vectors.append(model.wv[word])
  return np.mean(vectors,axis=0)

sentence = "nlp is interesting"
vec = sent_vec(sentence)
print(vec)
```

```python
# consine similarity score
from sklearn.metrics.pairwise import cosine_similarity

s1 = sent_vec("nlp is interesting")
s2 = sent_vec("machine learning is interesting")

similarity = cosine_similarity([s1],[s2])
print("sentence similarity\n")
print(similarity[0][0])
```
