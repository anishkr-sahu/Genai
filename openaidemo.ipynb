ChatCompletion API 
- it understand multiple prompts
- we can pass multiple prompts

```python
prompt = input("Enter the prompt: ")

response = openai.ChatCompletion.create(
    model = "gpt-3.5-turbo",
    messages = [
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user","content": prompt}
    ],
    # parameter
    max_token = 25, #  no of tokens
    temperatute = 0.6, # randomness to the response
    n = 2 # how many response you want
)

print(response["choices"][0]["message"]["content"])
```

Function calling in openAI
- it helps the openai model to interact with different third party api
- it creates a communication bridge between api and model ( feeding the data into model)

integrate different 3rd party
- websites
- database
- wikipedia
- html page
- docs


```python
# example dummy function hard coded to return the same weather
# in production, this could be your backend API or an external API

import requests

def get_current_weather(location):
    """ Get the current weather in a given location"""
    url = ""
    querystring = {"text": location}

    headers = {
        'x-rapidapi-key':"",
        'x-apidapi-host': ""
    }

    response = requests.get(url, headers=headers, params=querystring)

    print(response.json())
    return response.json()
```

```python
# response = get_current_weather("Delhi")
```

```python
functions = [
    {
        "name": "get_current_weather",
        "description": "Eget the current weather in a given location",
        "parameters": {
            "type": "object",
            "properties": {
                "location":{
                    "type": "string",
                    "description": "The city and state, e.g. new delhi"
                },
            },
            "required":["location"],
        },
    }
]
```

```python
user_message = "today temperature of delhi"
messages=[]
messages.append({"role": "user","content": user_message})

completion = openai.ChatCompletion.create(
    model = "gpt-3.5-turbo",
    messages = messages,
    functions=functions
)
```

```python
response=completion.choices[0].message
function_name = response['function_call']['name']
```

```python
import json
location = eval(response['function_call']['arguments'])['location']
```

```python
# send the info on the function call and function response to gpt
messages.append(response) # external
messages.append(
    {
        "role": "function",
        "name": function_name,
        "content": location
    }
)
```

```python
# initialize gpt model
second_response = openai.ChatCompletion.create(
    model = "gpt-3.5-turbo",
    messages=messages,
    functions=functions
)
```

```python
# response
print(second_response.choices[0].message)
```
